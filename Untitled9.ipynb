```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# display settings
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("superstore_level2.csv")
```

```python
df = df.drop_duplicates()
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].mean())
df['Profit'] = df['Profit'].fillna(0)
```

```python
df = df.dropna()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y', errors='coerce')
```

```python
df['Category'] = df['Category'].str.strip()
df['Region'] = df['Region'].str.strip()
```

```python
top_states = df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(5)
print(top_states)
```

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(category_profit)
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()
print(monthly_sales)
```

```python
sub_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
print(sub_profit.head())
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Sales', data=df)
plt.title("Sales by Category")
plt.show()
```

```python
monthly_sales.plot(kind='line', figsize=(10,5))
plt.title("Monthly Sales Trend")
plt.ylabel("Sales")
plt.show()
```

```python
region_sales.plot(kind='bar', figsize=(8,5))
plt.title("Region Wise Sales")
plt.show()
```

```python
top_sub = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
top_sub.plot(kind='bar')
plt.title("Profit by Sub Category")
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(df['Sales'], df['Profit'], alpha=0.6)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

 key Insight 


Technology category generates the highest sales revenue.

The Western region contributes the largest portion of total sales.

Phones and Chairs are among the most profitable sub-categories.

Sales show seasonal trends with certain months having higher sales.

Some sub-categories generate high sales but low profit, indicating high cost or discounting.

A few states dominate overall sales performance.
